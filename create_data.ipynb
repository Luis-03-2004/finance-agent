```python
import pandas as pd
import random
from datetime import datetime, timedelta
```

```python
# Categorias de gastos para simular a dor do cliente
categorias = ["Alimentação", "Transporte", "Moradia", "Lazer", "Juros Rotativo", "Pagamento Fatura"]
estabelecimentos = ["iFood", "Uber", "Mercado Livre", "Itaú Cartões", "Posto Ipiranga", "Netflix"]

# Gerando 100 transações falsas
dados = []
data_atual = datetime.now()

for i in range(20):
    categoria = random.choice(categorias)
    
    # Lógica simples para dar realismo aos dados
    if categoria == "Juros Rotativo":
        valor = round(random.uniform(50.0, 300.0), 2)
        estabelecimento = "Itaú Cartões"
    elif categoria == "Pagamento Fatura":
        valor = round(random.uniform(500.0, 2000.0), 2)
        estabelecimento = "Itaú Cartões"
    else:
        valor = round(random.uniform(15.0, 150.0), 2)
        estabelecimento = random.choice([e for e in estabelecimentos if e != "Itaú Cartões"])

    data_transacao = data_atual - timedelta(days=random.randint(0, 30))
    
    dados.append({
        "id_cliente": "C-12345",
        "data": data_transacao.strftime("%Y-%m-%d"),
        "estabelecimento": estabelecimento,
        "categoria": categoria,
        "valor_reais": valor,
        "tipo": "Saída"
    })

# Adicionando a entrada do salário para o Agente calcular quanto sobra
dados.append({
    "id_cliente": "C-12345",
    "data": (data_atual - timedelta(days=5)).strftime("%Y-%m-%d"),
    "estabelecimento": "AmoraDev Tecnologia",
    "categoria": "Salário",
    "valor_reais": 3000.00,
    "tipo": "Entrada"
})
```

```python

# Exportando para CSV
df = pd.DataFrame(dados)
df.to_csv("extrato_simulado_itau.csv", index=False)
print("Base de dados sintética gerada com sucesso! 🚀")
```
